```python
batsman = pd.read_csv('sharma-kohli.csv')
batsman.head(3)
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title("crickeater data")
plt.xlabel("year")
plt.ylabel("run score")
plt.show()
```

```python
# set colour graph

plt.plot(batsman['index'],batsman['V Kohli'],color='red')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')
plt.title("crickeater data")
plt.xlabel("year")
plt.ylabel("run score")
plt.show()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='red',label='viat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',label='shrma')
plt.title("crickeater data")
plt.xlabel("year")
plt.ylabel("run score")
plt.legend()
plt.show()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='brown')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6')
plt.title("batsman record two cricketar ")
plt.ylabel("run rate this information")
plt.xlabel("year this play ")
plt.grid()
```
