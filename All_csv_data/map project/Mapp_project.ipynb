```python
import numpy as np
import pandas as pd
import sys
import matplotlib
```

```python
census = pd.read_csv("/content/india-districts-census-2011.csv")
# census.head(2)
```

```python
cols=['District code','District name','Population','Male','Female','Literate','Households_with_Internet']
```

```python
census=census[cols]
census.head(2)
```

```python
latlong = pd.read_csv("/content/district wise centroids.csv")
latlong.head()
```

```python
final_df=latlong.merge(census,left_on='District',right_on='District name').drop(columns=['District name']).head()
final_df.head(3)
```

```python
# this is ratio of female
final_df['sex_ratio']=round((final_df['Female']/final_df['Male'])*100)
final_df.head(1)
```

```python
final_df['Literacy_ratio']=round((final_df['Literate']/final_df['Population'])*100)
final_df.head(2)
```

```python
final_df.drop(columns=['Male','Female','Literate'])
```

```python
final_df.to_csv('Mapp_cleen_data1',index=False)
```
